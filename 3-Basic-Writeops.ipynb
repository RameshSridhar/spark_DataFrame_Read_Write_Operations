### Extracting data into spark by creating dataframe

```python
df1=spark.read.csv("/Volumes/catalog_one/schema_one/volume_one/directory_one/custs_header", header=True, sep=",", inferSchema=True, samplingRatio=0.10)
```

### Schema migration, Data Conversion

```python
df2=df1.write.csv(path="/Volumes/catalog_one/schema_one/volume_one/Target_directory/custs_header_out", sep="~",header=True, mode="overwrite")
```

```python
transformed_df1=df1.select("custid","fname","lname","profession","age").withColumnRenamed("profession","prof")
transformed_df1.write.csv(path="/Volumes/catalog_one/schema_one/volume_one/Target_directory/custs_header_out", sep="~",header=True,mode="overwrite", compression="gzip")

transformed_df1.select("custid","prof","age").write.json(path="/Volumes/catalog_one/schema_one/volume_one/Target_directory/jsonout",mode="append")
display(spark.read.json("/Volumes/catalog_one/schema_one/volume_one/Target_directory/jsonout"))
```

```python
col_name="id int,first_name string, last_name string, age int, occupation string"
df1=spark.read.schema(col_name).csv("/Volumes/catalog_one/schema_one/volume_one/directory_one/custsmodified",sep=",")
#df1.show(10)
transformed_df1=df1.select("id","first_name","last_name","age","occupation").withColumnRenamed("occupation","Profession")
transformed_df1.write.csv(path="/Volumes/catalog_one/schema_one/volume_one/Target_directory/custsmodified_out",sep=",",header=True,mode="overwrite")
#display(spark.read.csv("/Volumes/catalog_one/schema_one/volume_one/Target_directory/custsmodified_out",header=True))
transformed_df1.write.json(path="/Volumes/catalog_one/schema_one/volume_one/Target_directory/json_out1",mode="overwrite")
display(spark.read.json("/Volumes/catalog_one/schema_one/volume_one/Target_directory/json_out1"))
df1.select("id","age","occupation").write.json(path="/Volumes/catalog_one/schema_one/volume_one/Target_directory/json_out1",mode="append")
display(spark.read.json("/Volumes/catalog_one/schema_one/volume_one/Target_directory/json_out1"))
```
